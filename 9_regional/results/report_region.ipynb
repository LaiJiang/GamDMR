# Whole Genome methylation Association Analysis for Allergic Asthma (AA): Smoothed Regional Analysis


We analyzed the methylation data with the region-based method, i.e. SOMNiBUS, to identify differentially methylated regions (DMRs) that are associated with Allergic Asthma.


The (supposed) advantages of SOMNiBUS:

* allowing non-linear, complex associations with multiple covariates using a Generalized Additive Model approach
* allowing smoothed regional association analysis of methylation data
* incoporate count-based methylation error rate arguments

## Method 


### Somnibus Model
Response: observed metylation count and total count of CpG sites in a region.

Covariate of interest: allergic asthma status (AA).

confounders: age, sex,...etc.



### Data Preparation

* Methylation proportions were transformed into a long format suitable for binomial regression, capturing both methylated and total read counts.

* Data was split into chunks with a maximum gap of 250 bp between sites. We select this threshold due to the observed distribution of cpg sites distribution:


![Figure 2. ](cpg_spacing.png)


* We conducted preliminary analyses and found out that at least 51 cpgs per region is required for this data to achieve a reasonable power for the SOMNIBUS model to converge.

* parameters tuning: We set the number of spline basis functions (n.k) adaptively based on region size (maximum of 3 or number of CpGs divided by 20).  We adopted the default measurement error rate of 0.1 for the SOMNiBUS model.

* Computational efficiency: To save computational time, we designed a two-stage analysis approach to cut the whole genome methylation set into regions and run the SOMNiBUS model on each region in parallel. 





## Results


### DMRs associated with AA status

We identified:

* 20 DMRs associated with AA status with pvalue < 1e-5. 
* 11 DMRs associated with AA status with pvalue < 5e-8.
* 19 DMRs associated with AA status with pvalue < 0.01 after multiple testing correction with Bonferroni method.


In the top 20 DMRs, only 2 DMRs are highly relevant with Asthma with existing literature evidence:

KLF4:   chr9:110249363-110252277
AHRR:   chr5:320573-323337

how about the previous famous gene PTPRE ("Genes to Diseases (G2D) Computational Method to Identify Asthma Candidate Genes") ?

It has pvalue 0.01, and ranked the 57th in the top 100 DMRs.

If we look at the top 100 DMRs and we found 5 DMRs that are highly relevant with Asthma with existing literature evidence:

* AHRR (Aryl-Hydrocarbon Receptor Repressor): associated with airway inflammation and asthma endotypes.

* PLCG1 (Phospholipase C Gamma 1):  T-cell receptor signaling, important in allergic asthma

* PTPRE (Protein Tyrosine Phosphatase, Receptor Type E)

* KLF4 (Kruppel-like Factor 4): epithelial barrier function, inflammation, and immune regulation — processes important in asthma pathogenesis

* WNT7B (Wnt Family Member 7B):  airway remodeling and inflammatory responses in asthma. 


There are also 5 DMRs indirect relevant with Asthma with existing literature evidence: MEST, FOXC2, ASCL2, SSTR1/SSTR4


### DMRs thats overlapping with Univariate analysis


only 1 cpg was found overlapping between SOMNiBUS and univariate analysis (Model 1 and Model 2):

10:99474180-99474181

This CpG is located in the gen body ADAM12, which is associated with cell adhesion, migration and tissue remodeling, andeven fibrotic lung diseases. 

There are no direct evidence of ADAM12 in asthma, but it is known that tissue remodeling genes have functional relevance in airway diesease.
