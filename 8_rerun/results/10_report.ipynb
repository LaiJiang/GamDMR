## Data 

Phenotype and genotype of interests:

* AA only versus control (neither AA nor AD)

* Methylation profile (methylation proportions at each CpG; after arc-sin square root transformation) from  Illumina 450k microarray.

Other Covariates:

* age at recruitment (AgeCalc)
* sex
* smoking status. Non-smoker versus smoker (current or former).
* Five surrogate variables (SVs) to capture hidden sources of variation in the methylation data. sv1, …sv5.
* Cell-type proportions.
* Random Effect: Family ID (FID) was included as a random effect to account for shared genetic and environmental factors within families.


## Models 


### Model 1

* Y: AA Status (0/1)

* X: arc-sin square root transformed methylation beta values at each CpG. 

* Other covariates as fixed effects.

* Family ID (FID) as random effect.

* algorithm: 
  - stage 1: feature selection using glmmLASSO with logit link.
  - stage 2: linear mixed effect model with the glmmLASSO selected features from stage 1 with logit link. Collect the final pvalue for the selected features.

* PTPRE

The second strongest signal.

Also found in Karine Tremblay and Catherine Laprise's paper "Genes to Diseases (G2D) Computational Method to Identify Asthma Candidate Genes" found rs7081735 in PTPRE associated with childhood allergic asthma in SLSJ cohort.

* CRHR1 

Multiple studies found that CRHR1 variants modulate inhaled corticosteroid (ICS) response in asthma (e.g. rs1876828 and rs242941). 


* VARS1

A very strong signal in the analysis. 

Located in HLA/MHC region, found associated with Neurodevelopmental Disorder With Microcephaly, Seizures, And Cortical Atrophy. But not direct evidence linking to asthma in existing studies.

### Conclusion


As a whole, univaraite assoication analysis between each CpG and allergic asthma status can inflate false positives (as a lone outlier drives the hit). This could explain some of the extreme low pvalues from model 1 and model 2, of which sometimes are due to perfect seperation between AA/control groups on odd CpGs. As a rsult, these scattered hits across the genome also make pathway enrichment analysis under-powered.

The single CpG analysis could also dilute true signals (if each individual site’s effect is small) by penalizing the pvalues too much.
