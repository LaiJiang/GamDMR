# Whole Genome methylation Association Analysis for Allergic Asthma (AA):  Aim 1


## Summary 

This report documents the results of whole genome methylation association analysis for allergic asthma (AA) under Aim 1. 


## Data 

Phenotype and genotype of interests:

* AA only versus control (neither AA nor AD)

* Methylation profile (methylation proportions at each CpG; after arc-sin square root transformation) from  Illumina 450k microarray.

Other Covariates:

* age at recruitment (AgeCalc)
* sex
* smoking status. Non-smoker versus smoker (current or former).
* Five surrogate variables (SVs) to capture hidden sources of variation in the methylation data. sv1, …sv5.
* Cell-type proportions.
* Random Effect: Family ID (FID) was included as a random effect to account for shared genetic and environmental factors within families.


## Models 


### Model 1

* Y: AA Status (0/1)

* X: arc-sin square root transformed methylation beta values at each CpG. 

* Other covariates as fixed effects.

* Family ID (FID) as random effect.

* algorithm: glmmLASSO with logit link.


* pros: 
  * 1. The fixed effects will be forced toward zero via optimized regularization, therefore the collinearity issue is resolved.
  * 2. The glmmLASSO algorithm is computationally efficient comparing to other penalization methods, making it suitable for whole genome analysis.

* cons:
    * 1. The choice of penalty parameter lambda can affect the results, and there is no universal method for selecting it. In our practise we select the lambda value by optimizing the model fit according to BIC criteria.
    * 3. Interpretation of the results can be challenging due to the complexity of the model.



### Model 2

* Y: arc-sin square root transformed methylation beta values at each CpG. 

* X: AA Status (0/1)  

* Other covariates as fixed effects.

* Family ID (FID) as random effect.

* algorithm: linear mixed effect model.


* pros: 
    * 1. The model is easy to interpret, as the coefficients represent the change in methylation associated with AA status.

* cons:
    * 1. The model assumes that the residuals are normally distributed and homoscedastic, which may not be universally true for all CpGs.
    * 2. An example is we observed that the model 2 will fail at CpGs where most methylation proportions are close to 0. We have to ignore these CpGs for model 2 analysis.
    * 3. The model can lead to overfitting and false positives for feature selection, since there are 13 fixed effects and 1 random effect in the model.
    * 4. The output indicated that for many CpGs the random effect for family had near-zero variance, suggesting little variability between families at those sites. In fact, around 10% of the top 10,733 CpGs are missing random effects! We have to ignore these CpGs for whole genome analysis.

## Why so dramatic differences?


### Model 1 vs Model 3

Model 1 (binary GLMM LASSO):

logit(Pr(AA)) ~ methylation (transformed) + ... + (1|FID)

We are asking "how well does methylation at each CpG predict AA status?"


Model 3: 

transformed methylation ~ AA status + ... + (1|FID)

Here we are asking the question "how much does AA status explain methylation at each CpG?"


Direction matters!

A CpG that is a strong predictor of AA status Do NOT  == strong "effect" of AA status on methylation of this CpG, and vice versa.



### Model 2 vs Model 3


Model 2:  an un-penalized linear mixed‐effects model. it uses likelihood‐ratio or Wald test for the AA status coefficient. Those tests yield classical p-values under the null distribution.

Model 3: 

* I calculated the pvalue from model using approximate likelihood estimation (i.e. Fisher-scoring refit) on the LASSO-selected coefficients.

* This is a problem due to shrinkage bias:

  * The penalty shrinks coefficients toward zero—often substantially—so even true signals would get its coefficient shrunk! This shrinkage inflates standard errors and therefore make p-values large.

  * In my post-analysis diagnosis, I found that for some CpGs with weak (but significiant) signal (in model 2), the model 3 would aggresively zero out the coefficients, and lead to high pvaues.


### Conclusion

The direction of the association + the shrunk effect lead to barren results of model 3.

Th current pvalue from model 1 and 3 is not valid.
